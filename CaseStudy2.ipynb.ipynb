```python
import pandas as pd

data = {
    "Student": ["S1", "S2", "S3", "S4", "S5",
                "S6", "S7", "S8", "S9", "S10"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 4, 6, 5],
    "Attendance": [75, 80, 85, 90, 95, 88, 92, 78, 84, 89],
    "Assignment_Score": [65, 70, 75, 80, 85, 88, 92, 72, 78, 83],
    "Previous_Score": [60, 65, 70, 75, 80, 85, 90, 68, 74, 79],
    "Final_Score": [62, 68, 73, 78, 84, 87, 91, 70, 77, 82]
}

df = pd.DataFrame(data)

print("Complete Dataset:")
print(df)

print("\nAverage Study Hours:")
print(df["Study_Hours"].mean())

print("\nAverage Attendance:")
print(df["Attendance"].mean())

print("\nAverage Assignment Score:")
print(df["Assignment_Score"].mean())

print("\nAverage Previous Score:")
print(df["Previous_Score"].mean())

print("\nAverage Final Score:")
print(df["Final_Score"].mean())

print("\nStudent with Highest Final Score:")
print(df[df["Final_Score"] == df["Final_Score"].max()])

print("\nStudent with Lowest Final Score:")
print(df[df["Final_Score"] == df["Final_Score"].min()])

print("\nStudents with Attendance Greater Than 80%:")
print(df[df["Attendance"] > 80])

print("\nStudents with Final Score Greater Than 80:")
print(df[df["Final_Score"] > 80])

print("\nStatistical Summary:")
print(df.describe())
```

```python
# Part C – NumPy Numerical Analysis

import numpy as np

final_scores = df["Final_Score"].to_numpy()

print("Final Scores:")
print(final_scores)

print("\nMean:")
print(np.mean(final_scores))

print("\nMedian:")
print(np.median(final_scores))

print("\nMaximum:")
print(np.max(final_scores))

print("\nMinimum:")
print(np.min(final_scores))

print("\nStandard Deviation:")
print(np.std(final_scores))
```

```python
#Part E – Linear Algebra
import numpy as np

X = df[[
    "Study_Hours",
    "Attendance",
    "Assignment_Score"
]].values
print("Matrix X:")
print(X)
print("\nShape of X:")
print(X.shape)
print("\nTranspose of X:")
print(X.T)
print("\nX Transpose X:")
print(X.T @ X)
study_hours = df["Study_Hours"].values
final_score = df["Final_Score"].values
print("\nDot Product:")
print(np.dot(study_hours, final_score))
```

```python
#Part F – Calculus: Differentiation
def F(x):
    return 40 + 5*x - 0.2*x**2
def dF(x):
    return 5 - 0.4*x
print("F(x) = 40 + 5x - 0.2x^2")
print("F'(x) = 5 - 0.4x")
x = 5 / 0.4
print("\nValue of x when F'(x) = 0:")
print(x)
print("\nValues:")
for x in range(1, 11):
    print("x =", x, "F(x) =", F(x), "F'(x) =", dF(x))
```

```python
#Part G – Calculus: Integration
import numpy as np
import matplotlib.pyplot as plt
def L(x):
    return 5*x - 0.2*x**2
integral = 2.5 * (5**2) - (0.2 / 3) * (5**3)
print("Integral from 0 to 5:")
print(integral)
x = np.linspace(0, 5, 1000)
y = L(x)
numerical_integral = np.trapezoid(y, x)
print("\nNumerical Approximation:")
print(numerical_integral)
plt.plot(x, y)
plt.xlabel("Study Hours")
plt.ylabel("Learning Benefit")
plt.title("Learning Benefit vs Study Hours")
plt.grid(True)
plt.show()
```

```python
#Part H – Correlation Analysis
import matplotlib.pyplot as plt
correlation = df.corr(numeric_only=True)
print("Correlation Matrix:")
print(correlation)
print("\nStudy Hours and Final Score:")
print(df["Study_Hours"].corr(df["Final_Score"]))
print("\nAttendance and Final Score:")
print(df["Attendance"].corr(df["Final_Score"]))
print("\nAssignment Score and Final Score:")
print(df["Assignment_Score"].corr(df["Final_Score"]))
print("\nPrevious Score and Final Score:")
print(df["Previous_Score"].corr(df["Final_Score"]))
final_correlation = correlation["Final_Score"].drop("Final_Score")
strongest = final_correlation.abs().idxmax()
print("\nStrongest Relationship:")
print(strongest)
print("\nCorrelation Value:")
print(final_correlation[strongest])
plt.figure(figsize=(8, 6))
plt.imshow(correlation, interpolation="nearest")
plt.colorbar()
plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)
plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
#Part I – Mini Prediction Model
import numpy as np
X = df[[
    "Study_Hours",
    "Attendance",
    "Assignment_Score"
]].values
Y = df["Final_Score"].values
X = np.column_stack((np.ones(len(X)), X))
beta = np.linalg.inv(
    X.T @ X
) @ X.T @ Y
print("Coefficients:")
print(beta)
print("\nIntercept:")
print(beta[0])
print("\nStudy Hours Coefficient:")
print(beta[1])
print("\nAttendance Coefficient:")
print(beta[2])
print("\nAssignment Score Coefficient:")
print(beta[3])
predicted_score = X @ beta
df["Predicted_Score"] = predicted_score
print("\nActual and Predicted Scores:")
print(df[[
    "Student",
    "Final_Score",
    "Predicted_Score"
]])
```
