```python
# SMART ELECTRIC VEHICLE BATTERY PERFORMANCE ANALYSIS
# Case Study Assignment
# Topics: Python Basics, NumPy, Linear Algebra and Calculus
import numpy as np

# TASK 1 - ANALYSIS USING PYTHON BASICS

print("=" * 60)
print("SMART ELECTRIC VEHICLE BATTERY PERFORMANCE ANALYSIS")
print("=" * 60)


# Battery sensor data
# Each reading contains:
# Time, Temperature, Voltage, Current, Charge

battery_data = [
    [0, 32, 400, 20, 95],
    [1, 35, 395, 22, 88],
    [2, 38, 390, 25, 80],
    [3, 40, 385, 28, 70],
    [4, 42, 380, 30, 60],
    [5, 45, 375, 32, 48]
]

# 1. Display all sensor readings

print("\n1. BATTERY SENSOR READINGS")
print("-" * 60)

print("Time(h)\tTemperature(°C)\tVoltage(V)\tCurrent(A)\tCharge(%)")

for reading in battery_data:
    print(f"{reading[0]}\t{reading[1]}\t\t\t{reading[2]}\t\t{reading[3]}\t\t{reading[4]}")

# 2. Identify high temperature
# Condition: T > 40°C

print("\n2. TEMPERATURE CHECK")
print("-" * 60)

for reading in battery_data:
    time = reading[0]
    temperature = reading[1]

    if temperature > 40:
        print(f"At {time} hour: High Temperature Alert")
    else:
        print(f"At {time} hour: Temperature Normal")

# 3. Identify low battery
# Condition: C < 50%

print("\n3. BATTERY CHARGE CHECK")
print("-" * 60)

for reading in battery_data:
    time = reading[0]
    charge = reading[4]

    if charge < 50:
        print(f"At {time} hour: Low Battery Alert")
    else:
        print(f"At {time} hour: Battery Charge Normal")

# 4. Function battery_status(charge)
# High   : charge > 70
# Medium : 50 <= charge <= 70
# Low    : charge < 50

def battery_status(charge):

    if charge > 70:
        return "High"
    elif charge >= 50:
        return "Medium"
    else:
        return "Low"


print("\n4. BATTERY STATUS")
print("-" * 60)

for reading in battery_data:
    time = reading[0]
    charge = reading[4]

    status = battery_status(charge)

    print(f"At {time} hour: Charge = {charge}% -> Status = {status}")

# TASK 2 - ANALYSIS USING NUMPY

print("\n" + "=" * 60)
print("TASK 2 - NUMPY ANALYSIS")
print("=" * 60)


# Convert sensor data into NumPy array
# Columns: Temperature, Voltage, Current, Charge

data = np.array([
    [32, 400, 20, 95],
    [35, 395, 22, 88],
    [38, 390, 25, 80],
    [40, 385, 28, 70],
    [42, 380, 30, 60],
    [45, 375, 32, 48]
])


# 1. Find shape of array

print("\n1. SHAPE OF ARRAY")
print("-" * 60)

rows, columns = data.shape

print("Data Matrix:")
print(data)

print("\nNumber of rows    :", rows)
print("Number of columns :", columns)
print("Shape             :", data.shape)

# 2. Calculate average of each parameter

print("\n2. AVERAGE OF EACH PARAMETER")
print("-" * 60)

average_temperature = np.mean(data[:, 0])
average_voltage = np.mean(data[:, 1])
average_current = np.mean(data[:, 2])
average_charge = np.mean(data[:, 3])

print("Average Temperature =", average_temperature, "°C")
print("Average Voltage     =", average_voltage, "V")
print("Average Current     =", average_current, "A")
print("Average Charge      =", average_charge, "%")


# 3. Maximum and minimum temperature

print("\n3. MAXIMUM AND MINIMUM TEMPERATURE")
print("-" * 60)

maximum_temperature = np.max(data[:, 0])
minimum_temperature = np.min(data[:, 0])

print("Maximum Temperature =", maximum_temperature, "°C")
print("Minimum Temperature =", minimum_temperature, "°C")

# 4. Extract voltage column

print("\n4. VOLTAGE COLUMN")
print("-" * 60)

voltage = data[:, 1]

print(voltage)

# 5. Identify high-temperature readings
# Condition: T > 40°C

print("\n5. HIGH-TEMPERATURE READINGS")
print("-" * 60)

high_temperature = data[data[:, 0] > 40]

print("Rows where temperature > 40°C:")
print(high_temperature)

# 6. Identify low battery readings
# Condition: C < 50%

print("\n6. LOW BATTERY READINGS")
print("-" * 60)

low_battery = data[data[:, 3] < 50]

print("Rows where charge < 50%:")
print(low_battery)

# TASK 3 - ANALYSIS USING LINEAR ALGEBRA


print("\n" + "=" * 60)
print("TASK 3 - LINEAR ALGEBRA")
print("=" * 60)


# Transformation matrix
#
# A = [ 2  1 ]
#     [ 1  3 ]

A = np.array([
    [2, 1],
    [1, 3]
])


# Sensor input
#
# X = [40]
#     [25]

X = np.array([
    [40],
    [25]
])


# 1. Matrix multiplication Y = AX

print("\n1. TRANSFORMED OUTPUT Y = AX")
print("-" * 60)

Y = np.matmul(A, X)

print("Matrix A:")
print(A)

print("\nInput X:")
print(X)

print("\nOutput Y = AX:")
print(Y)

# 2. Transpose of A

print("\n2. TRANSPOSE OF A")
print("-" * 60)

A_transpose = A.T

print("A^T =")
print(A_transpose)


# 3. Determinant of A

print("\n3. DETERMINANT OF A")
print("-" * 60)

det_A = np.linalg.det(A)

print("det(A) =", det_A)

# 4. Inverse of A

print("\n4. INVERSE OF A")
print("-" * 60)

if det_A != 0:
    A_inverse = np.linalg.inv(A)
    print("A^-1 =")
    print(A_inverse)
else:
    print("Inverse does not exist because determinant is zero.")

# 5. Solve system of equations
#
# 2x + y = 100
# x + 3y = 120


print("\n5. SOLVING SYSTEM OF EQUATIONS")
print("-" * 60)

A_system = np.array([
    [2, 1],
    [1, 3]
])

B_system = np.array([
    100,
    120
])


solution = np.linalg.solve(A_system, B_system)

x = solution[0]
y = solution[1]

print("Equations:")
print("2x + y = 100")
print("x + 3y = 120")

print("\nSolution:")
print("x =", x)
print("y =", y)

# TASK 4 - ANALYSIS USING CALCULUS

print("\n" + "=" * 60)
print("TASK 4 - CALCULUS ANALYSIS")
print("=" * 60)


# Battery charge function:
#
# C(t) = 100 - 2t - (1/2)t^2
#
# Derivative:
#
# dC/dt = -2 - t


# 1. Rate of change of battery charge

print("\n1. RATE OF CHANGE OF BATTERY CHARGE")
print("-" * 60)

print("Battery charge function:")
print("C(t) = 100 - 2t - (1/2)t^2")

print("\nDerivative:")
print("dC/dt = -2 - t")

# Create function for derivative

def rate_of_change(t):
    return -2 - t

# 2. Calculate discharge rate at t = 2 and t = 5

print("\n2. DISCHARGE RATE")
print("-" * 60)

rate_at_2 = rate_of_change(2)
rate_at_5 = rate_of_change(5)

print("dC/dt at t = 2 =", rate_at_2, "% per hour")
print("dC/dt at t = 5 =", rate_at_5, "% per hour")

# 3. Interpret the result

print("\n3. INTERPRETATION")
print("-" * 60)

print("At t = 2, battery charge is decreasing at 4% per hour.")
print("At t = 5, battery charge is decreasing at 7% per hour.")
print("Therefore, the battery is losing charge at an increasing rate.")

# FINAL BATTERY PERFORMANCE REPORT

print("\n" + "=" * 60)
print("FINAL BATTERY PERFORMANCE REPORT")
print("=" * 60)

# 1. Safety - first time temperature exceeds 40°C


print("\n1. SAFETY")
print("-" * 60)

for reading in battery_data:
    time = reading[0]
    temperature = reading[1]

    if temperature > 40:
        print("Temperature first exceeds 40°C at:", time, "hour")
        break

# 2. Total charge reduction

print("\n2. BATTERY CHARGE")
print("-" * 60)

initial_charge = battery_data[0][4]
final_charge = battery_data[-1][4]

charge_loss = initial_charge - final_charge

print("Initial Charge =", initial_charge, "%")
print("Final Charge   =", final_charge, "%")
print("Total Charge Loss =", charge_loss, "%")

# 3. NumPy analysis summary

print("\n3. NUMPY ANALYSIS SUMMARY")
print("-" * 60)

print("Average Temperature =", average_temperature, "°C")
print("Average Voltage     =", average_voltage, "V")
print("Average Current     =", average_current, "A")
print("Average Charge      =", average_charge, "%")

# 4. Linear Algebra summary

print("\n4. LINEAR ALGEBRA SUMMARY")
print("-" * 60)

print("Transformation Matrix A:")
print(A)

print("\nInput X:")
print(X)

print("\nY = AX:")
print(Y)

print("\nTranspose of A:")
print(A_transpose)

print("\nDeterminant of A =", det_A)

print("\nInverse of A:")
print(A_inverse)

print("\nSolution of AX = B:")
print("x =", x)
print("y =", y)

# 5. Calculus summary

print("\n5. CALCULUS SUMMARY")
print("-" * 60)

print("C(t) = 100 - 2t - (1/2)t^2")
print("dC/dt = -2 - t")

print("At t = 2: dC/dt =", rate_at_2, "% per hour")
print("At t = 5: dC/dt =", rate_at_5, "% per hour")

print("Conclusion: Battery is losing charge at an increasing rate.")

# 6. FINAL DECISION

print("\n6. FINAL DECISION")
print("-" * 60)

print("Decision: MONITOR")
print()
print("Reason:")
print("- Temperature exceeds 40°C at 4 hours.")
print("- Temperature reaches 45°C by the end of the test.")
print("- Battery charge falls from 95% to 48%.")
print("- Charge becomes less than 50% at 5 hours.")
print("- Discharge rate becomes faster with time.")
print("- Therefore, continuous monitoring is recommended.")


# FINAL DISCUSSION


print("\n" + "=" * 60)
print("FINAL DISCUSSION")
print("=" * 60)

print("""
Factors that may indicate unsafe battery operation:
1. Increasing battery temperature.
2. Temperature exceeding 40°C.
3. Low battery charge.
4. Increasing discharge rate.
5. Decreasing battery voltage.
6. Increasing current.

Additional data that can be collected:
1. Battery cell temperature.
2. State of Health (SOH).
3. State of Charge (SOC).
4. Internal resistance.
5. Charging and discharging cycles.
6. Individual cell voltage.
7. Cooling system performance.
8. Battery pressure and other safety sensor readings.
""")

print("=" * 60)
print("END OF BATTERY PERFORMANCE ANALYSIS")
print("=" * 60)
```
