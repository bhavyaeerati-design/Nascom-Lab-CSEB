```python
a = np.array([1, 2, 3])
b = np.array([4, 0, 1])
print('a . b  (dot)   :', np.dot(a, b))
print('a x b  (cross) :', np.cross(a, b))
```

```python
print('L2 norm  ||a||_2 :', la.norm(a))
print('L1 norm  ||a||_1 :', la.norm(a, 1))
print('Linf norm        :', la.norm(a, np.inf))
def cosine(u, v):
    return np.dot(u, v) / (la.norm(u) * la.norm(v))

print('cosine(a, b)     :', round(cosine(a, b), 3))
```

```python
A = np.array([[2., 1.],
              [1., 3.]])
B = np.array([[1., 0.],
              [4., 2.]])
print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n', A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))
```

```python
I = np.eye(2)
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))
print('A symmetric?', np.allclose(A, A.T))
theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)
S = np.array([[1.5, 0.0],
              [0.0, 0.5]])
t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])
scaled  = S @ square
rotated = R @ square
print('Rotated corners:\n', rotated)
```

```python
def close_loop(pts):
    return np.hstack([pts, pts[:, :1]])
fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
A = np.array([[2., 0.],
              [0., 3.]])
vals, vecs = la.eig(A)
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python
for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),    # pair 1
    (np.array([1, 2, 3]), np.array([3, 2, 1])),    # pair 2
    (np.array([2, 0, 0]), np.array([0, 5, 0])),    # pair 3
]

for i, (u, v) in enumerate(pairs, start=1):
    # 1. L1 and L2 norm of u
    l1 = np.linalg.norm(u, ord=1)
    l2 = np.linalg.norm(u, ord=2)

    # 2. Cosine similarity of u and v
    cosine_similarity = np.dot(u, v) / (
        np.linalg.norm(u) * np.linalg.norm(v)
    )

    print(f"\nPair {i}:")
    print("L1 norm of u =", l1)
    print("L2 norm of u =", l2)
    print("Cosine similarity =", cosine_similarity)

# 3. Most similar pair
print("\nMost similar pair = Pair 1")
```

```python
import numpy as np

M = np.array([[4., 2.],
              [2., 3.]])
P = np.array([[0., -1.],
              [1.,  0.]])   # a 90-degree rotation

# 1. Inverse and trace of M
inv_M = np.linalg.inv(M)
trace_M = np.trace(M)

print("Inverse of M:\n", inv_M)
print("Trace of M =", trace_M)

# 2. M @ inv(M) == I ?
I = np.eye(2)
print("\nM @ inv(M):\n", M @ inv_M)
print("Is M @ inv(M) equal to I?", np.allclose(M @ inv_M, I))

# 3. Is M symmetric? Is P orthogonal?
print("\nIs M symmetric?", np.array_equal(M, M.T))
print("Is P orthogonal?", np.allclose(P.T @ P, np.eye(2)))
```

```python
import numpy as np
import matplotlib.pyplot as plt

tri = np.array([[0, 1, 0.5],
                [0, 0, 1.0]])   # 3 corners of a triangle

# 1. Scaling matrix (x*2, y*0.5)
S = np.array([[2, 0],
              [0, 0.5]])

# 2. Rotation matrix (45 degrees)
theta = np.radians(45)

R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])

# 3. Apply scaling first, then rotation
transformed = R @ S @ tri

# Close the triangles for plotting
original_plot = np.hstack((tri, tri[:, [0]]))
transformed_plot = np.hstack((transformed, transformed[:, [0]]))

# Plot original and transformed triangles
plt.plot(original_plot[0], original_plot[1], 'o-', label='Original')
plt.plot(transformed_plot[0], transformed_plot[1], 's-', label='Transformed')

plt.axis('equal')
plt.grid()
plt.legend()
plt.title("Original vs Transformed Triangle")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

```python
C = np.array([[4., 1.],
              [2., 3.]])

# 1. Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(C)

# 2. Print eigenvalues
print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)

# 3. Verify C v = lambda v for the first eigenvector
v = eigenvectors[:, 0]
lam = eigenvalues[0]

print("\nFirst eigenvector v:")
print(v)

print("\nC @ v:")
print(C @ v)

print("\nlambda * v:")
print(lam * v)

print("\nVerification:")
print(np.allclose(C @ v, lam * v))
```

```python
import numpy as np

A2 = np.array([[3., 2.],
               [1., 4.]])
b2 = np.array([7., 9.])

cat = np.array([0.9, 0.8, 0.1])
dog = np.array([0.85, 0.7, 0.2])
car = np.array([0.1, 0.2, 0.95])

# 1. Solve A2 x = b2
x = np.linalg.solve(A2, b2)
print("Solution x =", x)

# 2. Rank of A2
rank = np.linalg.matrix_rank(A2)
print("Rank of A2 =", rank)

# 3. Cosine similarity
cos_cat_dog = np.dot(cat, dog) / (
    np.linalg.norm(cat) * np.linalg.norm(dog)
)

cos_cat_car = np.dot(cat, car) / (
    np.linalg.norm(cat) * np.linalg.norm(car)
)

print("Cosine(cat, dog) =", cos_cat_dog)
print("Cosine(cat, car) =", cos_cat_car)

if cos_cat_dog > cos_cat_car:
    print("More similar: cat and dog")
else:
    print("More similar: cat and car")
```
